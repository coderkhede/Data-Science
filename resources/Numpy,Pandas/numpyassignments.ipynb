```python
import numpy as np
structured_arry=np.array([('Jack',25,25),('Jill',22,55),('John',28,35)], dtype=[('Name','U10'),('Age','i4'),('Weight','i4')])
print(structured_arry)
add_data=np.array([('Jane',24,65)], dtype=[('Name','U10'),('Age','i4'),('Weight','i4')])
structured_arry=np.append(structured_arry, add_data)
print(structured_arry)
print(np.sort(structured_arry, order='Age'))
```

```python
dtype=[("x","i4"),("y","i4")]
arr=np.array([(1,2),(3,4)], dtype=dtype)
print(arr)
eudic_dis=np.sqrt((arr["x"][1]-arr["x"][0])**2+(arr["y"][1]-arr["y"][0])**2)
print(eudic_dis)
```

```python
mat=np.random.randint(1,26, size=(5,5))
print(mat)
mean=np.mean(mat)
sd=np.std(mat)
var=np.var(mat)
median=np.median(mat)
print("Mean:", mean)
print("Standard Deviation:", sd)
print("Variance:", var)
print("Median:", median)

mat1=np.arange(1,10).reshape(3,3)
print(mat1)
mean1=np.mean(mat1)
sd1=np.std(mat1)
Normalized_mat=(mat1-mean1)/sd1
print(Normalized_mat)
```

```python
mat1=np.random.randint(1,100,size=(3,4))
mat2=np.random.randint(1,100,size=(3,4))
print("Matrix 1:\n", mat1)
print("Matrix 2:\n", mat2)
added_mat=mat1+mat2
subtracted_mat=mat1-mat2
multiplied_mat=mat1*mat2
divided_mat=mat1/mat2
print("Added Matrix:\n", added_mat)
print("Subtracted Matrix:\n", subtracted_mat)
print("Multiplied Matrix:\n", multiplied_mat)
print("Divided Matrix:\n", divided_mat)
```

```python
mat=np.arange(1,17).reshape(4,4)
print(mat)
row_wise_sum=np.sum(mat, axis=1)
col_wise_sum=np.sum(mat, axis=0)
print("Row-wise Sum:", row_wise_sum)
print("Column-wise Sum:", col_wise_sum)
```

```python
mat=np.random.randint(1,30,size=(3,3))
print(mat)
vac=np.random.randint(1,30,size=(3,))
print("Vector:", vac)
broadcasted_sum=mat+vac
print("Broadcasted Sum:\n", broadcasted_sum)
mat1=np.random.randint(1,31,size=(4,4))
print(mat1)
vac1=np.random.randint(1,31,size=(4,))
print("Vector:", vac1)
broadcasted_sum1=mat1-vac1
print("Broadcasted Sum1:\n", broadcasted_sum1)
```

```python
matr=-np.random.randint(1,50,size=(3,3))
print(matr)
det=np.linalg.det(matr).round(2)
print("Determinant:", det)
inv=np.linalg.inv(matr).round(2)
print("Inverse:\n", inv)
egv=np.linalg.eigvals(matr).round(2)
print("Eigenvalues: \n", egv)
```

```python
mat1=np.random.randint(1,15,size=(2,3))
print(mat1)
mat2=np.random.randint(1,15,size=(3,2))
print(mat2)
product=np.dot(mat1, mat2)
print("Product of Matrices:\n", product)
cross_product=np.cross(mat1, mat2.T)
print("Cross Product of Matrices:\n", cross_product)
```
