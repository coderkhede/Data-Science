```python
# qestion 1
class Animal:
    def __init__(self,name,spiecies):
        self.name=name
        self.spiecies=spiecies
class Dog(Animal):
    def __init__(self,name,spiecies,breed):
        super().__init__(name,spiecies)
        self.breed=breed
dog=Dog("Buddy","Mammal","Golden Retriever")
print(dog.name, " Attribute: Name")  # Output: Buddy
print(dog.spiecies , " Attribute: Species")  # Output: Mammal
print(dog.breed, " Attribute: Breed")  # Output: Golden Retriever
```

```python
#question 2
class Animal:
    def __init__(self,name,spiecies):
        self.name=name
        self.spiecies=spiecies
    def __str__(self):
        return f"{self.name} is a {self.spiecies}"
class Dog(Animal):
    def __init__(self,name,spiecies,breed):
        super().__init__(name,spiecies)
        self.breed=breed
    def __str__(self):
        return super().__str__() + f" and belongs to the {self.breed} breed"
dog=Dog("Buddy","dOG","GOLDEN RETRIEVER")
print(dog)
```

```python
# question 4
class Walker:
    def __init__(self,name):
        self.name=name
    def walk(self):
        return f"{self.name} is Walking..."
class Runner:
    def __init__(self,name):
        self.name=name
    def run(self):
        return f"{self.name} is Running..."
class Athalete(Walker,Runner):
    def __init__(self,name):
        super().__init__(name)
athalete=Athalete("John")
print(athalete.walk())  # Output: John is Walking...
print(athalete.run())   # Output: John is Running...
```

```python
# question 5
class Walker:
    def __init__(self,name):
        self.name=name
    def walk(self):
        return f"{self.name} is Walking..."
class Runner:
    def __init__(self,name):
        self.name=name
    def run(self):
        return f"{self.name} is Running..."
class Athalete(Walker,Runner):
    def __init__(self,name):
        super().__init__(name)
    def walk(self):
        return super().walk() + " and is an Athalete"
athalete=Athalete("John")
print(athalete.walk())  # Output: John is Walking... and is an Athalete
print(athalete.run())   # Output: John is Running...
```

```python
# question 6
class Walker:
    def __init__(self,name):
        self.name=name
    def walk(self):
        return f"{self.name} is Walking..."
class Runner:
    def __init__(self,name):
        self.name=name
    def run(self):
        return f"{self.name} is Running..."
class Athalete(Walker,Runner):
    def __init__(self,name,trainning_hours):
        super().__init__(name)
        self.training_hours=trainning_hours
    def train(self):
        return f"{self.name} is training for {self.training_hours} hours."
athalete=Athalete("John",5)
print(athalete.train())  # Output: John is training for 5 hours.
```

```python
class A:
    def show (self):
        return "Method in class A"
class B(A):
    def show (self):
        return "Method in class B"
class C(A):
    def show (self):
        return "Method in class C"
class D(B,C):
    def show (self):
        return "Method in class D"
d=D()
print(d.show())  # Output: Method in class D
 
```

```python
# quetion 12
class Base1:
    def __init__(self,a):
        self.a=a
class Base2:
    def __init__(self,b):
        self.b=b 
class Derived(Base1,Base2):
    def __init__(self,a,b,c):
        super().__init__(a)
        Base2.__init__(self,b)
    
        self.c=c
derived=Derived(1,2,3)
print(derived.a)  # Output: 1
print(derived.b)  # Output: 2       
print(derived.c)  # Output: 3
```

```python
#question 13
class Animal:
    def __init__(self,name,spiecies):
        self.name=name
        self.spiecies=spiecies
class Cat(Animal):
    def __init__(self,name,spiecies,color):
        super().__init__(name,spiecies)
        self.color=color
cat=Cat("Whiskers","Mammal","Black")
dog=Animal("Buddy","Mammal")
isinstance_cat=isinstance(cat,Cat)
isinstance_dog=isinstance(dog,Cat)
print(isinstance_cat)  # Output: True
print(isinstance_dog)  # Output: True
```

```python
#question 14
class Bird:
    def speak(self):
        return "Chirp!"
class Parrot(Bird):
    def speak(self):
        return "Squawk!"
class Panguin(Bird):
    def speak(self):
        return "Honk!"
parrot=Parrot()
panguin=Panguin()
brd=Bird()
print(parrot.speak())  # Output: Squawk!
print(panguin.speak())  # Output: Honk!
print(brd.speak())  # Output: Chirp!
```

```python
### Assignment 15: Combining Single and Multiple Inheritance

''' Create a base class named `Device` with an attribute `brand`. Create a derived class `Phone` that
 inherits from `Device` and adds an attribute `model`. Create another base class `Camera` with an
  attribute `resolution`. Create a derived class `Smartphone` that inherits from both `Phone` and
  `Camera`. Create an object of the `Smartphone` class and print its attributes. '''

class Device:
    def __init__(self,brand):
        self.brand=brand
class Phone(Device):
    def __init__(self,brand,model):
        super().__init__(brand)
        self.model=model
class Camera:
    def __init__(self,resolution):
        self.resolution=resolution
class Smartphone(Phone,Camera):
    def __init__(self,brand,model,resolution):
        Phone.__init__(self,brand,model)
        Camera.__init__(self,resolution)
smartphone=Smartphone("Apple","iPhone 13","12MP")
print(f"Brand: {smartphone.brand}")  # Output: Brand: Apple
print(f"Model: {smartphone.model}")  # Output: Model: iPhone
print(f"Resolution: {smartphone.resolution}")  # Output: Resolution: 12MP
```
