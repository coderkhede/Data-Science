```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns 
%matplotlib inline
```

```python
df=pd.read_csv('Travel.csv')
df
```

```python
print(df['Gender'].value_counts())
print(df['MaritalStatus'].value_counts())
```

```python
df['Gender']=df['Gender'].replace('Fe Male','Female')
df['MaritalStatus']=df['MaritalStatus'].replace('Single','Unmarried')
```

```python
print(df['Gender'].value_counts())
print(df['MaritalStatus'].value_counts())
```

```python
df.isnull().sum()
```

```python
feature_with_na=[feature for feature in df.columns if df[feature].isnull().sum()>=1]
for feature in feature_with_na:
    print(feature,np.round(df[feature].isnull().mean()*100,5))
```

```python
df[feature_with_na].select_dtypes(exclude='object').describe()
```

```python
df['Age']=df['Age'].fillna(df['Age'].median())
df['DurationOfPitch']=df['DurationOfPitch'].fillna(df['DurationOfPitch'].mode()[0])
df['NumberOfFollowups']=df['NumberOfFollowups'].fillna(df['NumberOfFollowups'].median())
df['PreferredPropertyStar']=df['PreferredPropertyStar'].fillna(df['PreferredPropertyStar'].mode()[0])
df['NumberOfTrips']=df['NumberOfTrips'].fillna(df['NumberOfTrips'].median())
df['NumberOfChildrenVisiting']=df['NumberOfChildrenVisiting'].fillna(df['NumberOfChildrenVisiting'].median())
df['MonthlyIncome']=df['MonthlyIncome'].fillna(df['MonthlyIncome'].mode()[0])
df['TypeofContact']=df['TypeofContact'].fillna(df['TypeofContact'].mode()[0])
df.isnull().sum()
```

```python
df.drop('CustomerID',inplace=True,axis=1)
```

```python
df.head()
```

```python
df['TotalVisiting']=df['NumberOfPersonVisiting']+df['NumberOfChildrenVisiting']
df.drop(['NumberOfPersonVisiting','NumberOfChildrenVisiting'],inplace=True,axis=1)
```

```python
df.head()
```

```python
num_features=[feature for feature in df.columns if df[feature].dtype!='str' ]
print('Number of Numeric features :',len(num_features))
cat_features=[feature for feature in df.columns if df[feature].dtype=='str']
print('Number of Catagorial features :',len(cat_features))
dis_features=[feature for feature in num_features if len(df[feature].unique())<=25]
print('Number of Discrete features :',len(dis_features))
conti_feature=[feature for feature in num_features if feature not in dis_features]
print('Number of Continious features :',len(conti_feature))
```

```python
X=df.drop('ProdTaken',axis=1)
y=df['ProdTaken']
```

```python
num_feature=[feature for feature in X.columns if X[feature].dtype!='str']
cat_feature=[feature for feature in X.columns if X[feature].dtype=='str']
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=10)
X_test.shape,X_train.shape
```

```python
from sklearn.preprocessing import StandardScaler,OneHotEncoder
from sklearn.compose import ColumnTransformer
num_trans=StandardScaler()
cat_trans=OneHotEncoder(drop='first')
preprocessing=ColumnTransformer(
    [
        ('OneHotEncoder',cat_trans,cat_feature),
        ('StandardScaler',num_trans,num_feature)
    ]
)
X_train=preprocessing.fit_transform(X_train)
X_test=preprocessing.transform(X_test)
pd.DataFrame(X_train)
```

```python
from sklearn.ensemble import RandomForestClassifier,AdaBoostClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report,confusion_matrix,f1_score,accuracy_score,roc_auc_score
```

```python
models={
    'Decision Tree': DecisionTreeClassifier(),
    'Random Forest': RandomForestClassifier(),
    'Adaboost Model':AdaBoostClassifier()
}
for i in range(len(list(models))):
    model=list(models.values())[i]
    model.fit(X_train,y_train)
    y_pred=model.predict(X_test)
    print(list(models.keys())[i])
    print("Classification report",classification_report(y_test,y_pred))
    print("confustion matrix :",confusion_matrix(y_test,y_pred))
    print("accuracy_score :",accuracy_score(y_test,y_pred))
    print('f1_score :',f1_score(y_test,y_pred))
    print('Roc Auc Score :',roc_auc_score(y_test,y_pred))
    print("\n")
```

```python
gv_parms={
    'n_estimators':[100,200,300,400,500,1000],
    'learning_rate':[1,2,3,4,5,10]
}
gv_model=[
    ('rf',AdaBoostClassifier(),gv_parms)
]
gv_model
```

```python
from sklearn.model_selection import GridSearchCV
ht_parms={}
for name,model,params in gv_model:
    grid=GridSearchCV(model,param_grid=params,cv=5,verbose=3,refit=True)
    grid.fit(X_train,y_train)
    g_pred=grid.predict(X_test)
    print("Classification report",classification_report(y_test,g_pred))
    print("confustion matrix :",confusion_matrix(y_test,g_pred))
    print("accuracy_score :",accuracy_score(y_test,g_pred))
    print('f1_score :',f1_score(y_test,g_pred))
    print('Roc Auc Score :',roc_auc_score(y_test,g_pred))
    print("\n")
```
