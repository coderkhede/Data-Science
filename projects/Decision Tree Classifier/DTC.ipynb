```python
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
from sklearn.datasets import load_iris
iris=load_iris()
iris
```

```python
X=pd.DataFrame(iris['data'],columns=['sepal length','sepal width','petal length','petal width'])
y=iris['target']
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=10)
X_train
```

```python
from sklearn.tree import DecisionTreeClassifier
dtc=DecisionTreeClassifier()
dtc.fit(X_train,y_train)
y_pred=dtc.predict(X_test)
```

```python
from sklearn import tree
plt.figure(figsize=(15,10))
tree.plot_tree(dtc,filled=True)
```

```python
from sklearn.metrics import classification_report,confusion_matrix,accuracy_score
print(classification_report(y_test,y_pred))
print(confusion_matrix(y_test,y_pred))
print(accuracy_score(y_test,y_pred))
```

```python
#postrunning : we put the max_depth to 2 
dtc1=DecisionTreeClassifier(max_depth=2)
dtc1.fit(X_train,y_train)
y_pred1=dtc1.predict(X_test)
print(classification_report(y_test,y_pred1))
print(confusion_matrix(y_test,y_pred1))
print(accuracy_score(y_test,y_pred1))
```

```python
#pre-prunning
params={
    'criterion':['gini', 'entropy', 'log_loss'],
    'splitter':['best', 'random'],
    'max_features':['sqrt', 'log2'],
    'max_depth':[1,2,3,4]
}
params
```

```python
from sklearn.model_selection import GridSearchCV
grid=GridSearchCV(DecisionTreeClassifier(),param_grid=params,cv=5,verbose=3,refit=True,scoring='accuracy')
grid.fit(X_train,y_train)
g_pred=grid.predict(X_test)
print(grid.best_params_)
print(grid.best_score_)
```

```python
print(classification_report(y_test,g_pred))
print(confusion_matrix(y_test,g_pred))
print(accuracy_score(y_test,g_pred))
```
