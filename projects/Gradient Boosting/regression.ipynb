```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
df=pd.read_csv('cardekho_imputated.csv')
df
```

```python
df.isnull().sum()
```

```python
df.drop(['car_name','brand','Unnamed: 0'],axis=1,inplace=True)
df
```

```python
df['model'].unique()
```

```python
from sklearn.preprocessing import LabelEncoder,StandardScaler,OneHotEncoder
lb=LabelEncoder()
df['model']=lb.fit_transform(df['model'])
df['model'].value_counts()
```

```python
X=df.drop(['selling_price'],axis=1)
y=df['selling_price']
X
```

```python
num_features=X.select_dtypes(exclude='str').columns
cat_features=X.select_dtypes(include='str').columns
print(num_features)
print(cat_features)
```

```python
from sklearn.compose import ColumnTransformer
num_trans=StandardScaler()
cat_trans=OneHotEncoder(drop='first')
processers=ColumnTransformer([
        ('OneHotEncoder',cat_trans,cat_features),
        ('StandardScaler',num_trans,num_features)
    ],remainder='passthrough'
)
X=processers.fit_transform(X)
pd.DataFrame(X)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=10)
```

```python
from sklearn.ensemble import RandomForestRegressor,AdaBoostRegressor,GradientBoostingRegressor
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import mean_squared_error,mean_absolute_error,r2_score

def evaluate_score(test,pred):
    mse=mean_squared_error(test,pred)
    mae=mean_absolute_error(test,pred)
    r2=r2_score(test,pred)
    return mse,mae,r2
```

```python
models={
    'Random Forest':RandomForestRegressor(),
    'KNN':KNeighborsRegressor(),
    "Adaboost":AdaBoostRegressor(),
    'Gradient_boosting':GradientBoostingRegressor()
}
for i in range(len(list(models))):
    model=list(models.values())[i]
    model.fit(X_train,y_train)
    y_pred=model.predict(X_test)
    mse_test,mae_test,r2_test=evaluate_score(y_test,y_pred)
    print(list(models.keys())[i])
    print("Mse : ",mse_test)
    print("MAE : ",mae_test)
    print("R2_score : ",r2_test)
    print('\n')
```

```python
from sklearn.model_selection import GridSearchCV
knn_params={
    'algorithm':['auto', 'ball_tree', 'kd_tree', 'brute'],
    'n_neighbors':[5,6,7,8]
}
random_params={
    'n_estimators':[100,500,1000],
    'max_features':['sqrt', 'log2', None]
}
ada_boost_params={
    'n_estimators':[50,100,250,500,1000],
    'loss':['linear', 'square', 'exponential']
}


gradient_params={
    'n_estimators':[100,500,1000],
    'min_samples_split':[2,4,6,8,10],
    'max_depth':[2,3,4,5],
    'loss':['squared_error', 'absolute_error', 'huber', 'quantile']
}
combo=[
    ('knn',KNeighborsRegressor(),knn_params),
    ('random forest',RandomForestRegressor(),random_params),
    ('Ada Boost',AdaBoostRegressor(),ada_boost_params),
    ('Gradient Boosting',GradientBoostingRegressor(),gradient_params)
]
for name,model,params in combo:
    grid=GridSearchCV(model,param_grid=params,cv=5,verbose=2,refit=True,n_jobs=-1)
    grid.fit(X_train,y_train)
    y_pred=grid.predict(X_test)
    mse,mae,r2=evaluate_score(y_test,y_pred)
    print(name)
    print(mse)
    print(mae)
    print(r2)
    print(grid.best_params_)
    print("\n")
```
