```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline

X=6* np.random.rand(100,1) - 3
y=1.5*X**2 + X + 2 + np.random.randn(100,1)
```

```python
plt.scatter(X,y,color="y")
```

```python
#train_test_split
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=42)
```

```python
#linear regression
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(x_train,y_train)
plt.scatter(x_train,y_train,color="y")
plt.plot(x_train,model.predict(x_train),color="r")
model.score(x_test,y_test)
```

```python
#polynomial Transformation
from sklearn.preprocessing import PolynomialFeatures
poly=PolynomialFeatures(degree=2,include_bias=True)
x_train_poly=poly.fit_transform(x_train)
x_test_poly=poly.transform(x_test)
```

```python
poly_model=LinearRegression()
poly_model.fit(x_train_poly,y_train)
plt.scatter(x_train,y_train,color="y")
plt.scatter(x_train,poly_model.predict(x_train_poly),color='g')
poly_model.score(x_train_poly,y_train)
```

```python
X_new=np.linspace(-3,3,100).reshape(100,1)
x_new_poly=poly.transform(X_new)
plt.scatter(x_train,y_train,color="y",label="Training Data")
plt.scatter(x_test,y_test,color="g",label="Testing Data")
plt.plot(X_new,poly_model.predict(x_new_poly),'r',label="Polynomial Fit on unseen data")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.show()
```

```python
#for degree 3 
poly=PolynomialFeatures(degree=3,include_bias=True)
x_train_poly=poly.fit_transform(x_train)
x_test_poly=poly.transform(x_test)
poly_model=LinearRegression()
poly_model.fit(x_train_poly,y_train)
plt.scatter(x_train,y_train,color="y")
plt.scatter(x_train,poly_model.predict(x_train_poly),color='g')
plt.show()
poly_model.score(x_train_poly,y_train)

X_new=np.linspace(-3,3,100).reshape(100,1)
x_new_poly=poly.transform(X_new)
plt.scatter(x_train,y_train,color="y",label="Training Data")
plt.scatter(x_test,y_test,color="g",label="Testing Data")
plt.plot(X_new,poly_model.predict(x_new_poly),'r',label="Polynomial Fit on unseen data degree 3")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.show()
poly_model.score(x_train_poly,y_train)
```
