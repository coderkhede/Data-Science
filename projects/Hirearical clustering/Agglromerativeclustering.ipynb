```python
import pandas as pd
import matplotlib.pyplot as plt
import scipy.cluster.hierarchy as sch
%matplotlib inline
```

```python
from sklearn.datasets import load_iris
iris=load_iris()
iris_data=pd.DataFrame(iris.data,columns=iris.feature_names)
iris_data
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
iris_data_scaled=scaler.fit_transform(iris_data)
from sklearn.decomposition import PCA
pca=PCA(n_components=2)
pca_data=pca.fit_transform(iris_data_scaled)
pca_data
```

```python
plt.Figure(figsize=(10,8))
plt.scatter(pca_data[:,0],pca_data[:,1],c=iris.target)
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.show()
```

```python
plt.figure(figsize=(8,6))
plt.title('Dendrogram')
sch.dendrogram(sch.linkage(pca_data,method='ward',metric='euclidean'))
plt.xlabel('Samples')
plt.ylabel('Euclidean distances')
plt.show()
```

```python
from sklearn.cluster import AgglomerativeClustering
ac=AgglomerativeClustering(n_clusters=2,metric='euclidean',linkage='ward')
ac.fit(pca_data)
ac.labels_
```

```python
plt.figure(figsize=(5,3))
plt.scatter(pca_data[:,0],pca_data[:,1],c=ac.labels_)
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('Agglomerative Clustering Results')
plt.show()
```
